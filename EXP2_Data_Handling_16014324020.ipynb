## 5. Handling missing values (imputation, using inplace)
Filling with **subdivision-wise mean** rather than a single global mean, since rainfall varies hugely by region.

```python
month_cols = ['JAN','FEB','MAR','APR','MAY','JUN','JUL','AUG','SEP','OCT','NOV','DEC']
season_cols = ['ANNUAL','JF','MAM','JJAS','OND']

# note: df[col].fillna(val, inplace=True) does NOT work anymore in newer pandas
# (copy-on-write breaks chained assignment) - reassigning df[col] is the fix.
for col in month_cols + season_cols:
    df[col] = df[col].fillna(df.groupby('SUBDIVISION')[col].transform('mean'))

# any subdivision still fully NaN for a column -> fallback to global mean, true inplace on whole df
df.fillna(df[month_cols + season_cols].mean(numeric_only=True), inplace=True)

print("Nulls remaining after imputation:", df.isnull().sum().sum())
```

## 9. Accessing rows on a condition, few columns (3 cases, compound conditions)

```python
print("Case 1 - Konkan & Goa, years after 2010:")
c1 = df[(df['SUBDIVISION'] == 'Konkan & Goa') & (df['YEAR'] > 2010)]
c1[['YEAR','ANNUAL']]
```

```python
print("Case 2 - ANNUAL > 3000mm AND JJAS > 2000mm:")
c2 = df[(df['ANNUAL'] > 3000) & (df['JJAS'] > 2000)]
c2[['SUBDIVISION','YEAR','ANNUAL','JJAS']].head(8)
```

```python
print("Case 3 - Coastal Karnataka OR Kerala, ANNUAL < 2000mm (low-rainfall year):")
c3 = df[((df['SUBDIVISION'] == 'Coastal Karnataka') | (df['SUBDIVISION'] == 'Kerala')) & (df['ANNUAL'] < 2000)]
c3[['SUBDIVISION','YEAR','ANNUAL']]
```

```python
print("Using numpy directly:")
print("max:", np.max(df['ANNUAL'].values), " min:", np.min(df['ANNUAL'].values))
```

## 11. Groupby on one / more columns (2 cases)

```python
print("Case 1 - groupby SUBDIVISION, mean ANNUAL rainfall:")
g1 = df.groupby('SUBDIVISION')['ANNUAL'].mean().sort_values(ascending=False)
g1.head(10)
```

```python
print("Case 2 - groupby [SUBDIVISION, DECADE], mean JJAS rainfall:")
g2 = df.groupby(['SUBDIVISION','DECADE'])['JJAS'].mean()
g2.head(15)
```

## 12. New column derived from existing columns

```python
# irrigation-advisory relevant flag: what share of annual rainfall is from the monsoon?
df['MONSOON_SHARE_PCT'] = (df['JJAS'] / df['ANNUAL']) * 100
df[['SUBDIVISION','YEAR','ANNUAL','JJAS','MONSOON_SHARE_PCT']].head()
```

## 13. Aggregate functions with groupby (2 cases)

```python
print("Case 1 - agg() with multiple stats on ANNUAL, per SUBDIVISION:")
agg1 = df.groupby('SUBDIVISION')['ANNUAL'].agg(['mean','min','max','std'])
agg1.head(10)
```

```python
print("Case 2 - agg() on multiple columns at once:")
agg2 = df.groupby('SUBDIVISION').agg({'ANNUAL':'mean', 'JJAS':'mean', 'MONSOON_SHARE_PCT':'mean'})
agg2.head(10)
```

```python
print("Subdivisions whose average ANNUAL rainfall > 3000mm:")
high_rain = g1[g1 > 3000]
high_rain
```

## 15. Correlation between two columns

```python
corr_val = df['JJAS'].corr(df['ANNUAL'])
print(f"Correlation between JJAS (monsoon) rainfall and ANNUAL rainfall: {corr_val:.4f}")
```

```python
print("Full correlation matrix of monthly + seasonal columns:")
df[month_cols + season_cols].corr().round(2)
```

## 16. Transformation - normalization (Min-Max scaling)

```python
df['ANNUAL_NORM'] = (df['ANNUAL'] - df['ANNUAL'].min()) / (df['ANNUAL'].max() - df['ANNUAL'].min())
df[['SUBDIVISION','YEAR','ANNUAL','ANNUAL_NORM']].head()
```

```python
print("Range check after normalization -> min:", df['ANNUAL_NORM'].min(), " max:", df['ANNUAL_NORM'].max())
```

## 17. Joining, merging, concatenation

```python
# small lookup table for a merge demo
state_lookup = pd.DataFrame({
    'SUBDIVISION': ['Konkan & Goa', 'Kerala', 'Coastal Karnataka', 'Gangetic West Bengal'],
    'REGION_TYPE': ['Coastal', 'Coastal', 'Coastal', 'Gangetic Plain']
})
merged = pd.merge(df, state_lookup, on='SUBDIVISION', how='left')
print("Merged dataframe (only matched rows shown):")
merged[merged['REGION_TYPE'].notnull()][['SUBDIVISION','YEAR','REGION_TYPE']].head()
```

```python
print("Concat demo (first 3 + last 3 rows):")
concat_demo = pd.concat([df.head(3), df.tail(3)])
concat_demo[['SUBDIVISION','YEAR','ANNUAL']]
```
